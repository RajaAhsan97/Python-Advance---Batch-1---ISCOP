## Conditional Statements

if, elif, else

```python
marks = int(input("Enter Marks"))

if marks >= 50:  # True
    print("Congratulations")
    print("Pass")
else:
    print("Failed")
    print("AP ko course repeat krna hoga")
```

```python
marks = 95

if marks >= 90: # True
    print("Grade A")
if marks >= 80:   # True
    print('Grade B')
if marks < 80:     # False
    print("Grade C")
```

```python
# if-elif-else

marks = 79

if marks >= 90 and marks <= 100: # False
    print("Grade A")
elif marks >= 80 and marks < 90: # True
    print("Grade B")
else:
    print("Grade C")
```

# Functions

1. function name shoul be written in SNake case
2. function can have parameters (input)
3. function can compute a result
4. function can return a result

```python
# function definition
def test_func():
    # Doc String
    """
        This is test function, it take no input and it does not return a value.
        It can only print the message
    """
    print("Hello! I am in function")
```

```python
test_func()
```

```python
print(test_func.__doc__)
```

```python
# function with parameters and return
def my_func(fname, lname):
    """
        Parameters:
            fname: first_name
            lname: last_name

        Return:
            combine first and last name
    """
    # String concatenation
    full_name = fname + ' ' + lname   

    return full_name
```

```python
name = my_func("Raja", "Ahsan")

print(name)
```

```python
# simple calculator function
def calculator_func(x, y, op):
    """
        This is a simple calculator function, it takes two value and an operator,
        and compute the result.

        Parameters:
            x: Value1
            y: Value2
            op: Arithmetic Operator (+, -, *, /)

        Return:
            It returns the computed result
    """
    if op == '+':
        result = x + y
    elif op == '-':
        result = x - y
    elif op == '*':
        result = x * y
    elif op == '/':
        if y == 0:
            raise ValueError("Can not divide by zero")
        result = x / y
    else:
        raise ValueError("Invalid operators, Use +,-,*,/")

    return result
```

```python
calculator_func()
```

```python
print(calculator_func.__doc__)
```

```python
r = calculator_func(2, 4, '/')

print(r)
```

## function with multiple return values

```python
def circle(r):
    """
        Compute area and circumference of a circle

        Parameter:
            r: radius

        Return:
            area and circumference
    """
    area = 3.14 * r ** 2
    circumference = 2 * 3.14 * r
    
    return area, circumference
```

```python
circle(5)
```

```python
circle_area, circle_circum = circle(5)

print(circle_area)
print(circle_circum)
```

### Scope of variables

```python
variable = 8    # Global variable

def func():
    variable = 9    # local variable

    print(variable)

print(variable)
func()
print(variable)
```

```python
# changing global variable value inside a function
variable = 8    # Global variable

def func():
    global variable
    variable = 9    # local variable

    print(variable)

print(variable)
func()
print(variable)
```

```python
variable = 40    # global variable

def outer_function():
    variable = 30    # local 
    def inner_function():
        variable = 10
        print(variable)   # enclosing variable
    print(variable)
    inner_function()
```

```python
outer_function()
```

#### Function parameters mapping

```python
def func(a,b,c):
    print(a,b,c)

# Positional mapping
func(1,2,3)

# Keyword mapping
func(c=3, a=1, b=2)

# Hybrid mapping
func(1, c=3, b=2)
```

### Lambda function

lambda arguments : expresssion

```python
# single argument
# square of number
square = lambda x : x ** 2 

print(square(5))
```

```python
# multiple argument
addition = lambda x, y : x + y

print(addition(5,6))
```

```python
# check even number
check_even = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check_even(9))
```

### Function with *args and **kwargs

```python
def func(*args):
    print(args)
    print("a=", args[0])
    print("b=", args[1])
    print("c=", args[2])

func(1,2,3)
```

```python
def func(**kwargs):
    print(kwargs)

func(name="Ahsan", experience=5, profession="freelance")
```
