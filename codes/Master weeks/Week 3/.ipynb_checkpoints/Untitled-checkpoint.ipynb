# 6. map()

The map() function in Python is a built-in tool that lets you apply a specific function to every item in an iterable (like a list, tuple, or dictionary) without using an explicit for loop.

### SYNTAX:

### map(function, iterable, ...)

```python
# Convert a list of strings into integers
string_numbers = ["1", "2", "3", "4"]
numbers = list(map(int, string_numbers))

numbers
```

```python
# Square every number in a list
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))

print(squared)
```

```python
# Add elements of two lists together side-by-side
list_a = [1, 2, 3]
list_b = [10, 20, 30]

sums = list(map(lambda x, y: x + y, list_a, list_b))

print(sums) 
```

```python
# removing whitespaces from string

s = ['  hello  ', '  world ', ' python  ']
res = map(str.strip, s)
print(list(res))
```

```python
# convert temperature in Celsius to Fahrenheit

celsius = [0, 20, 37, 100]
fahrenheit = map(lambda c: (c * 9/5) + 32, celsius)
print(list(fahrenheit))
```

# 7. filter()

The built-in filter() function in Python extracts elements from an iterable (like a list, tuple, or set) that satisfy a specific condition

### SYNTAX:

### filter(function, iterable)

```python
# Extract numbers greater than 10
numbers = [2, 15, 8, 23, 4, 42]

def is_greater_than_10(num):
    return num > 10

# Convert the filter object back into a list
result = list(filter(is_greater_than_10, numbers))
print(result) 
```

```python
# Keep only the even numbers
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```

# 8. reduce()

The reduce() function in Python applies a specific function cumulatively to the elements of an iterable (like a list) to reduce the entire sequence down to a single value

### SYNTAX:

### reduce(function, iterable, initializer=None)

```python
from functools import reduce
```

```python
# Summing list of numbers

numbers = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, numbers)

print(result)
```

```python
# finding maximum value

numbers = [15, 42, 12, 90, 8]
max_val = reduce(lambda x, y: x if x > y else y, numbers)

print(max_val)
```
