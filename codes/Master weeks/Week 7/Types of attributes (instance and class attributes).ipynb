Types of Methods:

Instance Methods: a. Defined with ---> def method_name(self)
b. First argument is self (refers to instance/object) b. Can access and modify instance attributes
Magic Methods: a. Special methods in Python (surrounded by __) b. Used for customized built-in operations like printing, addition, length etc

## Class Attributes

```python
class StudentRecords():
    # class attributes (shared across all objects/instances)
    institute = "Saylani Mass IT"
    
    # Part1: defining class and instance attributes
    # construtor
    def __init__(self, name, roll_no, course, marks):
        # instance attributes/variables
        self.name = name   #create an instance attribute name and assign it the value
        self.roll_no = roll_no
        self.course = course
        self.marks = marks

    # Part2: 
    # When we print the object/instance, then the memory address where object is created is printed.
    # We use __str__ method to define what get displayed when we print the object
    def __str__(self):    # magic method
        return f"Student: {self.name} enrolled in course: {self.course}"

    # Part3:
    # Instance methods - Behaviour of class on instance attributes
    def display(self):
        print(f"Name: {self.name}\nRoll no: {self.roll_no}\nCourse: {self.course}\n Marks: {self.marks}\n Institute: {self.institute}")

    # Part4:
    # Instance methods
    def ispass(self):
        if self.marks > 50:
            #self.is_pass = True
            return f"Student: {self.name} passes"        
        else:
            #self.is_pass = False
            return f"Student: {self.name} fails"
```

```python
# creating objects from class
student1 = StudentRecords("Ahsan", "qw321", "AI and DS", 99)
student2 = StudentRecords("Taha", "we422", "AI and DS", 88)
```

## Static methods: ko class aur instance bi call kesakty h, es ko object nai chaiey hota
## Class Methods: ko class sy bi call krwa skty hein aur object sy bi call krwa sakty hein
## Instance methods: ko call keny k liey object chaiey hota h

```python
class StudentsRecords():
    # class attributes
    institute = "SMIT"
    campus = "Hajyani campus"
    
    # constructor
    def __init__(self, nm, r_no, course, marks):
        # instance attributes
        self.name = nm             # Public
        self.roll_no = r_no        # Public
        self.course = course       # Public
        self.__marks = marks       # Private
  
    # Magic method
    def __str__(self):
        return f"Student {self.name} bearing {self.roll_no} is enrolled in the course {self.course}"

    # instance method
    def display(self):
        print(f"------- Institute Details --------")
        print(f"Institute: {self.institute}")
        print(f"Campus: {self.campus}")
        print(f"------- Student details ----------")
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"Course: {self.course}")
        print(f"Marks: {self.__marks}")

    # getter method (To access private attributes)   -- instance method
    def get_marks(self):
        return self.__marks

    # setter method (To modify private attributes)   -- instance method
    def update_marks(self, marks):
        self.__marks = marks


    # display grade method -- instance method
    def display_grade(self):
        grade = StudentsRecords.convert_marks_to_grades(self.__marks)
        print(f"Grade: {grade}")

    # Class method
    @classmethod
    def get_institute(cls):
        return cls.institute
    
    # Static method --> helper method
    @staticmethod
    def convert_marks_to_grades(marks):
        print("------ SMIT grading system -------")
        print("Criteria for Quiz")
        print("Pass if marks >= 70%   --> A Grade")
        print("Fails if marks < 70%   --> B Grade") 

        if marks >= 70:
            return "A"
        else:
            return "B"
```

```python
student1 = StudentsRecords("Faheem", "qw321", "AI and DS", 99)

print(student1)
print(student1.institute)
print(student1.campus)
print(student1.display())
```
